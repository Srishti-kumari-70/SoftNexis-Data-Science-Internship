**STEP 1: Import Libraries & Load Dataset**

```python

# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset directly from Colab
df = pd.read_csv('/content/retail_store_sales.csv')

# Display dataset information
print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

**Step 2: Initial Data Inspection**

```python

# STEP 2: Initial Data Inspection

# 1. Dataset structure
print("Dataset Information:")
df.info()

# 2. Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all").T)

# 3. Missing values analysis
missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

print("\nMissing Values Report:")
display(missing_report)

# 4. Duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())

# 5. Unique values in categorical columns
print("\nUnique Values:")
for col in df.select_dtypes(include="object").columns:
    print(f"{col}: {df[col].nunique()} unique values")

# 6. Memory usage
memory = df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"\nMemory Usage: {memory:.2f} MB")
```

**STEP 3 — Deduplication**

```python

# STEP 3: Deduplication

# 1. Check exact duplicate rows
print("Exact Duplicate Rows:", df.duplicated().sum())

# 2. Check duplicate Transaction IDs
print(
    "Duplicate Transaction IDs:",
    df["Transaction ID"].duplicated().sum()
)

# 3. Check duplicate Customer IDs
print(
    "Unique Customers:",
    df["Customer ID"].nunique()
)

# 4. Remove exact duplicate rows
df = df.drop_duplicates().copy()

# 5. Verify after removal
print("\nAfter Deduplication:")
print("Total Rows:", df.shape[0])
print("Remaining Duplicates:", df.duplicated().sum())

# Validation
assert df.duplicated().sum() == 0

print("\nDeduplication completed successfully!")
```

**STEP 4 — Column Management**

```python

# STEP 4: Column Management

# Rename columns
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# Arrange columns logically
column_order = [
    "transaction_id",
    "customer_id",
    "transaction_date",
    "category",
    "item",
    "price_per_unit",
    "quantity",
    "total_spent",
    "payment_method",
    "location",
    "discount_applied"
]

df = df[column_order]

# Display updated columns
print("Updated Column Names:")
print(df.columns.tolist())

print("\nUpdated Dataset:")
display(df.head())
```

**STEP 4: Missing Value Handling**

```python
# 1. Fill missing item names
df['item'] = df['item'].fillna('Unknown')

# 2. Fill missing prices using category-wise median
df['price_per_unit'] = df['price_per_unit'].fillna(
    df.groupby('category')['price_per_unit'].transform('median')
)

# 3. Fill missing quantity using category-wise median
df['quantity'] = df['quantity'].fillna(
    df.groupby('category')['quantity'].transform('median')
)

# 4. Calculate missing total spent
df['total_spent'] = df['total_spent'].fillna(
    df['price_per_unit'] * df['quantity']
)

# 5. Preserve unknown discount information
df['discount_applied'] = df['discount_applied'].fillna('Unknown')

# Display missing values after cleaning
print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("\nUpdated Dataset:")
display(df.head())
```

**STEP 5: Data Type Correction**

```python
# 1. Convert transaction_date into datetime
df['transaction_date'] = pd.to_datetime(
    df['transaction_date'], errors='coerce'
)

# 2. Convert numerical columns into numeric format
numeric_cols = ['price_per_unit', 'quantity', 'total_spent']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 3. Convert discount_applied into categorical format
df['discount_applied'] = df['discount_applied'].astype('category')

# 4. Convert text columns into string format
text_cols = [
    'transaction_id', 'customer_id', 'category',
    'item', 'payment_method', 'location'
]

for col in text_cols:
    df[col] = df[col].astype('string')

# Display updated data types
print("Updated Data Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

**STEP 6: Format Standardization**

```python
# 1. Remove extra spaces and standardize category names
df['category'] = (
    df['category']
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
    .str.title()
)

# 2. Standardize item names
df['item'] = (
    df['item']
    .str.strip()
    .str.upper()
)

# 3. Standardize payment methods
df['payment_method'] = (
    df['payment_method']
    .str.strip()
    .str.title()
)

# 4. Standardize location
df['location'] = (
    df['location']
    .str.strip()
    .str.title()
)

# 5. Standardize transaction and customer IDs
df['transaction_id'] = df['transaction_id'].str.strip().str.upper()
df['customer_id'] = df['customer_id'].str.strip().str.upper()

# 6. Display unique values after standardization
for col in ['category', 'payment_method', 'location']:
    print(f"\n{col}:")
    print(df[col].unique())

print("\nFormat Standardization Completed!")
```

**Step 7: Outlier Detection**

```python
numeric_cols = ['price_per_unit', 'quantity', 'total_spent']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\nColumn: {col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers Found:", len(outliers))

print("\nOutlier Detection Completed!")
```

**Step 8: Final Data Validation**

```python
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nDuplicate Transaction IDs:",
      df['transaction_id'].duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nFinal Dataset Preview:")
display(df.head())
```

**Step 9: Export Cleaned Dataset**

```python

# STEP 9: Export Cleaned Dataset

output_file = '/content/cleaned_retail_store_sales.csv'

df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_file)

# Download the cleaned CSV
from google.colab import files
files.download(output_file)
```
